```python
# ------------------------------------------------------
# 1. CARGA DEL DATASET Y LIBRERÍAS
# ------------------------------------------------------
```

```python
import os
import json
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
)
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
# 1Detectar si el entorno de ejecución actual está dentro de la carpeta 'notebook'
if os.path.basename(os.getcwd()) == "notebook":
    os.chdir("..")  # Subir un nivel hacia la raíz del proyecto

# Confirmar en consola cuál es el directorio raíz actual
print(f"Directorio de trabajo configurado en: {os.getcwd()}")
```

```python
# Crear la estructura de carpetas solicitada
directorios = [
    "./data/raw/",
    "./data/processed/",
    "./outputs/",
    "./imagenes/",
    "./referencias/",
]

for d in directorios:
    os.makedirs(d, exist_ok=True)
```

```python
# Carga de datos
df_raw = pd.read_csv("data/raw/dataset_original.csv")
print(f"Dataset cargado correctamente.")
print("Dimensiones iniciales del dataset:", df_raw.shape)
```

```python
# ------------------------------------------------------
# 2. REVISIÓN DE COLUMNAS
# ------------------------------------------------------
```

```python
# Inspeccionar la estructura general, nombres exactos de variables y tipos de datos detectados.
print("\n--- Estructura y nombres de columnas originales ---")
print(df_raw.info())
print("\nPrimeras 5 filas del datasetl:")
print(df_raw.head())
```

```python
# ------------------------------------------------------
# 3. REVISIÓN DE NULOS
# ------------------------------------------------------
```

```python
# En los datasets de INEGI, las cifras confidenciales o no disponibles en localidades pequeñas se marcan con '*'.
cols_interes = ["VPH_C_ELEC", "VPH_INTER", "VPH_CEL", "VPH_STVP", "VPH_PC"]

print("\n--- Conteo de datos confidenciales ('*') por columna ---")
for col in cols_interes:
    conteo = (df_raw[col] == "*").sum()
    print(f"Columna {col}: {conteo} registros con '*'")

print("\n--- Conteo de valores nulos nativos (NaN) ---")
print(df_raw[cols_interes].isnull().sum())
```

```python
# ------------------------------------------------------
# 4. LIMPIEZA BÁSICA
# ------------------------------------------------------
```

```python
# Excluir los agregados estatales y municipales (MUN==0 o LOC agregada) para quedarnos solo con localidades reales.
df_limpio = df_raw[
    (df_raw["MUN"] > 0) & (df_raw["LOC"] > 0) & (df_raw["LOC"] < 9998)
].copy()
```

```python
# Convertir las columnas a tipo numérico; los '*' se convierten a NaN y luego se imputan con 0 (sin equipamiento).
for col in cols_interes:
    df_limpio[col] = pd.to_numeric(df_limpio[col], errors="coerce").fillna(0)
```

```python
# Feature engineering: calcular el porcentaje de viviendas con internet respecto a las que tienen electricidad.
df_limpio["pct_internet"] = np.where(
    df_limpio["VPH_C_ELEC"] > 0,
    (df_limpio["VPH_INTER"] / df_limpio["VPH_C_ELEC"]) * 100,
    0,
)

print(f"\nRegistros procesados tras filtrado de localidades: {df_limpio.shape[0]}")
```

```python
# Crear la variable objetivo binaria (1: Alta conectividad >= 30%, 0: Baja/Media < 30%).
df_limpio["conectividad_alta"] = (df_limpio["pct_internet"] >= 30.0).astype(
    int
)

df_limpio.to_csv("data/processed/dataset_limpio.csv", index=False)
```

```python
# ------------------------------------------------------
# 5. SELECCIÓN DE VARIABLES
# ------------------------------------------------------
```

```python
# Definir las variables predictoras (X) y la variable objetivo (y).
features_num = ["VPH_C_ELEC", "VPH_CEL", "VPH_STVP", "VPH_PC"]
feature_cat = ["NOM_MUN"]

X = df_limpio[features_num + feature_cat]
y = df_limpio["conectividad_alta"]
```

```python
# Convertir la variable categórica municipal a variables indicadoras numéricas (One-Hot Encoding).
X_encoded = pd.get_dummies(X, columns=feature_cat, drop_first=False)

print("\nDistribución de clases de la variable objetivo (conectividad_alta):")
print(y.value_counts(normalize=True))
```

```python
# ------------------------------------------------------
# 6. ANÁLISIS EXPLORATORIO
# ------------------------------------------------------
```

```python
plt.figure(figsize=(12, 4))

# Gráfico 1: Histograma de la distribución del porcentaje de viviendas con internet
plt.subplot(1, 2, 1)
sns.histplot(
    df_limpio["pct_internet"], bins=30, kde=True, color="skyblue"
)
plt.axvline(
    30, color="red", linestyle="--", label="Umbral Conectividad Alta (30%)"
)
plt.title("Distribución del % de Viviendas con Internet")
plt.xlabel("% de Internet")
plt.ylabel("Localidades")
plt.legend()
plt.savefig("imagenes/grafica_1.png")
plt.close()

# Gráfico 2: Conteo de la variable objetivo binaria
plt.subplot(1, 2, 2)
sns.countplot(
    x="conectividad_alta",
    data=df_limpio,
    palette="Set2",
    hue="conectividad_alta",
    legend=False,
)
plt.title("Conteo de Clases (0: Baja/Media, 1: Alta)")
plt.xlabel("Clase Conectividad")
plt.ylabel("Número de Localidades")
plt.tight_layout()
plt.savefig("imagenes/grafica_2.png")
plt.close()

plt.show()
```

```python
# ------------------------------------------------------
# 7. MODELO APLICADO
# ------------------------------------------------------
```

```python
# Dividir los datos en entrenamiento (70%) y prueba (30%) estratificando la variable objetivo.
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.3, random_state=42, stratify=y
)

# Inicializar y entrenar el algoritmo DecisionTreeClassifier.
modelo = DecisionTreeClassifier(
    max_depth=4, random_state=42, criterion="gini"
)
modelo.fit(X_train, y_train)

print(f"\nModelo entrenado con éxito. Muestras en conjunto de entrenamiento: {X_train.shape[0]}")
```

```python
# ------------------------------------------------------
# 8. RESULTADOS
# ------------------------------------------------------
```

```python
# Generar las predicciones con el conjunto de prueba.
y_pred = modelo.predict(X_test)

# Guardar outputs
test_results = X_test.copy()
test_results["conectividad_alta_real"] = y_test
test_results["prediccion_conectividad_alta"] = y_pred
test_results.to_csv("outputs/resultados_modelo.csv", index=False)

# Evaluar la exactitud y generamos el reporte de métricas de clasificación.
acc = accuracy_score(y_test, y_pred)
print(f"\nExactitud (Accuracy) global del modelo: {acc:.4f}")
print("\nReporte detallado de clasificación:")
print(classification_report(y_test, y_pred))
```

```python
# ------------------------------------------------------
# 9. GRÁFICAS O TABLAS
# ------------------------------------------------------
```

```python
plt.figure(figsize=(12, 5))

# Matriz de confusión
plt.subplot(1, 2, 1)
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Baja/Media", "Alta"],
    yticklabels=["Baja/Media", "Alta"],
)
plt.title("Matriz de Confusión")
plt.xlabel("Predicción")

plt.ylabel("Realidad")

# Importancia de las variables predictoras (Top 6)
plt.subplot(1, 2, 2)
importancias = pd.Series(
    modelo.feature_importances_, index=X_encoded.columns
)
top_importancias = importancias.nlargest(6)
top_importancias.plot(kind="barh", color="teal")
plt.title("Top 6 Variables Predictoras Más Importantes")
plt.xlabel("Importancia Relativa")

plt.tight_layout()
plt.show()

# Representación visual de la estructura del Árbol de Decisión
plt.figure(figsize=(16, 7))
plot_tree(
    modelo,
    feature_names=X_encoded.columns,
    class_names=["Baja/Media", "Alta"],
    filled=True,
    rounded=True,
    fontsize=8,
)
plt.title("Estructura del Árbol de Decisión Entrenado")
plt.show()
```

```python
# ------------------------------------------------------
# 10. INTERPRETACIÓN EN CELDAS MARKDOWN
# ------------------------------------------------------
```

### Interpretación del Modelo de Clasificación

1. **Rendimiento General:**
   * El modelo alcanzó una **exactitud del 77.22%** en el conjunto de prueba, lo que sugiere una capacidad razonable para clasificar el nivel de conectividad por localidad. La clase Baja/Media obtuvo mayor precision (0.86) mientras que la clase Alta mostró mayor recall (0.84).

2. **Variables Clave:**
   * La disponibilidad de **Computadoras / Laptops (`VPH_PC`)** y **Teléfonos Celulares (`VPH_CEL`)** resultaron ser las variables con mayor peso predictivo dentro del árbol de decisión. Con los datos disponibles, se observa que estas variables están fuertemente asociadas con el acceso a internet.

3. **Aplicación Práctica:**
   * Los resultados sugieren que el modelo podría servir como herramienta exploratoria para identificar zonas con baja conectividad y orientar la inversión en infraestructura de telecomunicaciones. Sin embargo, este resultado debe interpretarse con cuidado, ya que el modelo no demuestra causalidad entre las variables.

4. **Limitaciones:**
   * La etiqueta `conectividad_alta` fue creada con un umbral arbitrario del 30%, lo cual no representa una verdad absoluta. Los datos corresponden a un periodo censal específico y las cifras confidenciales fueron imputadas con 0, lo que puede subestimar valores reales en localidades pequeñas.

```python
# Exportar resumen de resultados por clase
report = classification_report(y_test, y_pred, output_dict=True)
resumen = pd.DataFrame(report).transpose()
resumen.to_csv("outputs/resumen_resultados.csv")
print("\nResumen exportado a outputs/resumen_resultados.csv")
```
